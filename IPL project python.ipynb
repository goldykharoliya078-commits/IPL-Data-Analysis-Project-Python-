```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load dataset
matches=pd.read_excel(r"C:\Users\BHAVISHYA\OneDrive\Desktop\matches.csv.xlsx")
matches
```

```python
deliveries=pd.read_excel(r"C:\Users\BHAVISHYA\OneDrive\Desktop\deliveries.csv.xlsx")
deliveries
```

```python
# find top teams
Top_teams=matches['winner'].value_counts().head(5)
Top_teams
```

```python
#find top batsman
top_batsman=deliveries.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(5)
top_batsman
```

```python
#find bowler
top_bowler=deliveries['bowler'].value_counts().head(5)
top_bowler
```

```python
top_teams=matches['winner'].value_counts().head(5)
top_teams
```

```python
#graph create
top_teams.plot(kind='bar', color=['red' , 'pink' , 'blue', 'purple', 'orange'])
plt.title('Top IPL Teams')
plt.xlabel('Team Name')
plt.ylabel('Number of Wins')
plt.show()
```
